```python
import pandas as pd
import numpy as np
import joblib

print("Libraries loaded successfully!")
```

```python
class SimpleKNNClassifier:
    
    def __init__(self, k=5):
        self.k = k
        self.X_train = None
        self.y_train = None
    
    def fit(self, X, y):
        self.X_train = X
        self.y_train = y
        return self
    
    def predict(self, X):
        predictions = []
        
        for sample in X:
            distances = np.sqrt(
                np.sum((self.X_train - sample) ** 2, axis=1)
            )
            
            nearest_indices = np.argsort(distances)[:self.k]
            nearest_labels = self.y_train[nearest_indices]
            
            values, counts = np.unique(
                nearest_labels,
                return_counts=True
            )
            
            prediction = values[np.argmax(counts)]
            predictions.append(prediction)
        
        return np.array(predictions)
```

```python
loaded_package = joblib.load("wastage_risk_model.joblib")

print("Saved model loaded successfully!")
print("Stored features:", len(loaded_package["features"]))
```

```python
df = pd.read_csv("fresh_produce_wastage_dataset.csv")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
```

```python
features = loaded_package["features"]

print("Features loaded from saved model:")
print(features)
print("Number of features:", len(features))
```

```python
sample = df[features].iloc[[0]].copy()

sample = sample.fillna(
    df[features].median(numeric_only=True)
)

sample_values = sample.to_numpy(dtype=float)

print("Sample prepared successfully!")
print("Sample shape:", sample_values.shape)
```

```python
sample_scaled = (
    sample_values - loaded_package["train_mean"]
) / loaded_package["train_std"]

print("Sample scaling completed!")
```

```python
sample_prediction = loaded_package["model"].predict(sample_scaled)

print("Prediction generated successfully!")
print("Predicted class number:", sample_prediction[0])
```

```python
predicted_class = loaded_package["number_to_label"][
    sample_prediction[0]
]

print("Predicted Wastage Risk Category:", predicted_class)
```
